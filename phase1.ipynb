## creating array from list 

```python
arr_1d=np.array([1,2,3,4,5])
print("1D Array:", arr_1d)


arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D Array: \n",arr_2d)
```

## Python list vs Numpy array


```python
py_list=[1,2,3]
print("python list multiplication :", py_list * 2)

np_arr = np.array([4,5,6])    # element wise multiplication
print("Python arry multiplication :", np_arr * 2)

import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\nlist operation time:",time.time()-start)

start = time.time()
np_array= np.arange(100000) *2
print("Numpy operation time:",time.time()-start)
```

## creating array from scratch

```python
zeros = np.zeros((3,4))
print("Zeros array: \n", zeros)

ones = np.ones((2,3))
print("ones array: \n",ones)


full = np.full((2,2),7)
print("full array: \n", full)

random = np.random.random((2,3))
print("random array: \n",random)


sequence  = np.arange(0,10,2)
print("sequence array: \n", sequence)
```

## Vector, Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector: ", vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("Matrix: \n",matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("Tensor: \n",tensor)
```

## Array properties

```python
arr = np.array([[1,2,3],
                [4,5,6]])
print("Shape",arr.shape)
print("dimension ",arr.ndim)
print("Size ",arr.size)
print("Data type ",arr.dtype)
```

## Array reshaping

```python
arr = np.arange(12)
print("Original array: ", arr)

reshaped = arr.reshape((3,4))       #converts this array into (3,4)matrix
print("Reshaped array: \n",reshaped)

flattened = reshaped.flatten()      #flatten the matrix array 
print("Flatten Array: \n", flattened)

raveled = reshaped.ravel()          #ravel (returns view, instead of copy)
print("Raveled Array: \n",raveled)

transpose = reshaped.T            #transposes the array 
print("Transposed Array: \n",transpose)

```
