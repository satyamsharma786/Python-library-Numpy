# Advance operations with business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("====Zomato Sales Analysis====")
print("\n Sales data shape:",sales_data.shape)
print("\n Sample data for 1st 3 restaurants:\n",sales_data[:3])
```

```python
# total sales per year 
print(np.sum(sales_data, axis=0))

yearly_total=np.sum(sales_data[:,1:],axis = 0)
print("yearly:\n",yearly_total)

# minimum sales per restaurant
min_sales= np.min(sales_data[:,1:], axis= 1)
print("\nminimum per restaurant:\n",min_sales)

# maximum sales per year
max_sales= np.max(sales_data[:,1:],axis = 0)
print("\nMax sales per year:\n",max_sales)

# average sales per restaurant
avg_sales= np.mean(sales_data[:,1:],axis=1)
print("\navg sales per restaurant:\n",avg_sales)

#cumulitive sales per restaurant
cumsum = np.cumsum(sales_data[:,1:],axis = 1)
print("\ncumulitive sales per restaurant:\n",cumsum)

plt.figure(figsize=(8,6))                                          #set figure size of the graphs 
plt.plot(np.mean(cumsum,axis=0))                                    #plot mean and cumulative sales on y axis 
plt.title("Average cumulative sales across all restaurant ")        #gives the title to the graph
plt.xlabel("Years")                                                 #gives title to the x axis
plt.ylabel("Sales")                                                 #gives title to the y axis
plt.grid(True)                                                      #decides to draw grid lines in the graph or not 
plt.show()                                                          #shows the graph 
```

# Operations related to vectors

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("vector addition:", vector1+vector2)
print("\n vector multiplication:", vector1 * vector2)
print("\n Dot product:", np.dot(vector1, vector2))
angle = np.arccos(np.dot(vector1, vector2)/ (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("Angle between vectors = ",angle)
```
